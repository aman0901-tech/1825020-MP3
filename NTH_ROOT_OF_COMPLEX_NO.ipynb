```python
import numpy as np
import cmath
import matplotlib.pyplot as plt

z = complex(float(input("Enter real part: ")),
            float(input("Enter imaginary part: ")))

n = int(input("Enter value of n: "))

r = abs(z)
theta = cmath.phase(z)

print("\nPolar Form of Complex Number:")
print("Modulus =", r)
print("Argument =", theta, "radians")

print("\nNth Roots:")

roots = []

for k in range(n):
    angle = (theta + 2 * np.pi * k) / n

    root = r**(1/n) * complex(np.cos(angle), np.sin(angle))

    roots.append(root)

    print("\nRoot", k+1)
    print("Cartesian Form =", root)
    print("Polar Form =", r**(1/n), "∠", angle)

plt.figure(figsize=(7, 7))

for i, root in enumerate(roots):
    plt.scatter(root.real, root.imag, s=80)
    plt.text(root.real, root.imag, f"  Root {i+1}")

plt.axhline(0)
plt.axvline(0)
plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("Nth Roots of a Complex Number")
plt.grid()
plt.axis("equal")
plt.show()


```
