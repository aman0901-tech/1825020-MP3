```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 10)
y = np.linspace(-2, 2, 10)

X, Y = np.meshgrid(x, y)

Z = X + 1j*Y

W = Z**2

print("Complex Mapping: W = Z^2")

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.scatter(X, Y)
plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("Z Plane")
plt.grid()

plt.subplot(1, 2, 2)
plt.scatter(W.real, W.imag)
plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("W Plane")
plt.grid()

plt.tight_layout()
plt.show()
```
