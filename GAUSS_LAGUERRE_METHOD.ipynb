```python
import numpy as np

def f(x):
    return x*x

n = int(input("Enter number of points: "))

x, w = np.polynomial.laguerre.laggauss(n)

sum = 0

for i in range(n):
    sum = sum + w[i]*f(x[i])

print("Integration =", sum)
```
