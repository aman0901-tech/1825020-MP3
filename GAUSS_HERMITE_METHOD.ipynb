```python
import numpy as np
import math

def f(x):
    return x*x

n = int(input("Enter number of points: "))

x, w = np.polynomial.hermite.hermgauss(n)

sum = 0

for i in range(n):
    sum = sum + w[i]*f(x[i])

print("Integration =", sum)

print("Exact Value =", math.sqrt(math.pi)/2)
```
