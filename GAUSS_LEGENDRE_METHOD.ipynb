```python
import numpy as np

def f(x):
    return x*x

a = float(input("Enter lower limit: "))
b = float(input("Enter upper limit: "))
n = int(input("Enter number of points: "))

x, w = np.polynomial.legendre.leggauss(n)

sum = 0

for i in range(n):
    t = ((b-a)*x[i] + (a+b))/2
    sum = sum + w[i]*f(t)

ans = (b-a)*sum/2

print("Integration =", ans)
```
