```python
import numpy as np

n = int(input("Enter number of data points: "))

x = []

for i in range(n):
    a = float(input("Enter value: "))
    x.append(a)

X = []

for k in range(n):
    s = 0
    for j in range(n):
        s += x[j] * np.exp(-2j*np.pi*k*j/n)
    X.append(s)

print("DFT is:")

for i in range(n):
    print(X[i])
```
