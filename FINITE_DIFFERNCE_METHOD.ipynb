```python
import numpy as np
import matplotlib.pyplot as plt

n = int(input("Enter number of intervals: "))

h = 1/n

A = np.zeros((n-1, n-1))
B = np.ones(n-1)*h**2

for i in range(n-1):
    A[i][i] = -2

    if i > 0:
        A[i][i-1] = 1

    if i < n-2:
        A[i][i+1] = 1

y = np.linalg.solve(A, B)

x = np.linspace(0, 1, n+1)

Y = np.zeros(n+1)
Y[1:n] = y

print("x values:", x)
print("y values:", Y)

plt.plot(x, Y, 'o-')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Finite Difference Method")
plt.grid()
plt.show()
```
