```python
import matplotlib.pyplot as plt
import math

x = float(input("Enter real part: "))
y = float(input("Enter imaginary part: "))

print("1. Translation")
print("2. Scaling")
print("3. Rotation")
print("4. Reflection")

ch = int(input("Enter your choice: "))

if ch == 1:
    a = float(input("Enter translation in x: "))
    b = float(input("Enter translation in y: "))

    u = x + a
    v = y + b

elif ch == 2:
    s = float(input("Enter scaling factor: "))

    u = x*s
    v = y*s

elif ch == 3:
    angle = float(input("Enter angle in degrees: "))

    angle = math.radians(angle)

    u = x*math.cos(angle) - y*math.sin(angle)
    v = x*math.sin(angle) + y*math.cos(angle)

elif ch == 4:
    u = x
    v = -y

else:
    print("Wrong choice")
    exit()

print("Original Point =", x, y)
print("Transformed Point =", u, v)

plt.quiver(0, 0, x, y, angles='xy',
           scale_units='xy', scale=1, color='blue',
           label='Original')

plt.quiver(0, 0, u, v, angles='xy',
           scale_units='xy', scale=1, color='red',
           label='Transformed')

plt.xlabel("Real Axis")
plt.ylabel("Imaginary Axis")
plt.title("2D Vector Transformation")
plt.grid()
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.axis("equal")
plt.show()







```
