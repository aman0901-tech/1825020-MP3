```python
import cmath
import math
import matplotlib.pyplot as plt

a = float(input("Enter real part: "))
b = float(input("Enter imaginary part: "))
n = int(input("Enter nth root: "))

z = complex(a, b)

r = abs(z)
theta = cmath.phase(z)

print("Original Number =", z)
print("Modulus =", r)
print("Argument =", theta)

x = []
y = []

for k in range(n):
    angle = (theta + 2*math.pi*k)/n

    real = r**(1/n)*math.cos(angle)
    imag = r**(1/n)*math.sin(angle)

    root = complex(real, imag)

    print("Root", k+1, "=", root)
    print("Cartesian Form =", real, "+", imag, "i")
    print("Polar Form =", r**(1/n), "angle", angle)

    x.append(real)
    y.append(imag)

plt.scatter(x, y)
plt.axhline(0)
plt.axvline(0)
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.title("Nth Roots of Complex Number")
plt.grid()
plt.show()
```
