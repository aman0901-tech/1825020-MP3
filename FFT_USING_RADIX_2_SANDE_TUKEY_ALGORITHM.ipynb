```python
import numpy as np

def fft(x):
    n = len(x)

    if n == 1:
        return x

    even = fft(x[0::2])
    odd = fft(x[1::2])

    X = [0]*n

    for k in range(n//2):
        t = np.exp(-2j*np.pi*k/n)*odd[k]
        X[k] = even[k] + t
        X[k+n//2] = even[k] - t

    return X

n = int(input("Enter number of points: "))

if n < 1 or (n & (n-1)) != 0:
    print("Enter power of 2 number of points")

else:
    x = []

    for i in range(n):
        x.append(float(input("Enter value: ")))

    result = fft(x)

    print("FFT is:")

    for i in result:
        print(i)
```
