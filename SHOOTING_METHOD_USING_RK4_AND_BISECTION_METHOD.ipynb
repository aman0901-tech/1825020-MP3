```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y, z):
    return z

def g(x, y, z):
    return -y

def rk4(s):
    h = np.pi/2/100
    x = 0
    y = 0
    z = s

    X = [x]
    Y = [y]

    for i in range(100):
        k1 = h*f(x, y, z)
        l1 = h*g(x, y, z)

        k2 = h*f(x+h/2, y+k1/2, z+l1/2)
        l2 = h*g(x+h/2, y+k1/2, z+l1/2)

        k3 = h*f(x+h/2, y+k2/2, z+l2/2)
        l3 = h*g(x+h/2, y+k2/2, z+l2/2)

        k4 = h*f(x+h, y+k3, z+l3)
        l4 = h*g(x+h, y+k3, z+l3)

        y += (k1+2*k2+2*k3+k4)/6
        z += (l1+2*l2+2*l3+l4)/6

        x += h

        X.append(x)
        Y.append(y)

    return Y[-1], X, Y

a = 0
b = 2

for i in range(30):
    c = (a+b)/2

    ya, _, _ = rk4(a)
    yc, _, _ = rk4(c)

    if ya*yc < 1:
        b = c
    else:
        a = c

value, X, Y = rk4(c)

print("Initial slope =", c)
print("Final value of y =", value)

plt.plot(X, Y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Shooting Method using RK4")
plt.grid()
plt.show()
```
